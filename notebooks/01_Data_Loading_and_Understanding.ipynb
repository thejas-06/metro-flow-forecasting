But boarding has fewer rows — lets check which stations have the mismatch.

## Investigating the Row Count Difference

```python
board_counts = df_board.groupby('Station').size().rename('Boarding_Rows')
exit_counts = df_exit.groupby('Station').size().rename('Exit_Rows')
```

```python
comp = pd.concat([board_counts, exit_counts], axis=1)
comp['Diff'] = comp['Exit_Rows'] - comp['Boarding_Rows']
comp[comp['Diff'] != 0]
```

15 stations have fewer boarding rows than exit rows. All of these are on the Yellow Line 
(R.V. Road to Bommasandra corridor).  

This happens in phased metro rollouts — exit gates record all passengers who deboard, 
but boarding (entry) may be restricted or not logged during certain dates when the line is 
under testing or partial operations.  

We will handle this during preprocessing by doing an outer merge and filling missing boarding counts with 0.

## Summary

**What we know so far:**
- Two datasets: boarding (entries) and exits — both from BMRCL Namma Metro, Bengaluru
- 83 stations, hourly granularity, Aug-Sep 2025
- No null values, no duplicates in either dataset
- Boarding has 92,280 rows, Exits has 95,616 rows (3,336 row gap from Yellow Line stations)
- ~33.8 million total passenger boardings recorded
- ~20% of rows are zero (late night hours when metro doesnt run)
- Semicolon delimiter in the raw CSV (not comma) — regional government data formatting
